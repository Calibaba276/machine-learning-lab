```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
file = pd.read_csv("./datasets/canada_per_capita_income.csv")
file.head()
```

```python
reg = linear_model.LinearRegression()

reg.fit(file[['year']], file['per capita income (US$)'])
```

```python
x = file['year']
y = file['per capita income (US$)']

m = reg.coef_
b = reg.intercept_

x_train = np.linspace(x.min(), x.max(), 100)
y_train = (m * x_train) + b

plt.scatter(x,y, label="Data Points")
plt.plot(x_train, y_train, label="Regression Line", color="gold")
plt.xlabel('Year')
plt.ylabel('Per Capita Income')
plt.legend()
```

```python
reg.predict([[2020]])
```
