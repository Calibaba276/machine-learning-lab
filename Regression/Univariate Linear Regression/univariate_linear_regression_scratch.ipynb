```python
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([5,7,9,11,13])
```

```python
def gradient_descent(x,y):
    m_curr = b_curr = 0
    iterations = 1000
    learning_rate = 0.1
    m = len(x)

    for i in range(iterations):
        # Default the m and b parameters to be 0
        y_predicted = m_curr * x + b_curr

        # Calculate cost to see how well the model is doing
        cost = (1/(2 * m)) * np.sum((y_predicted - y) ** 2)

        # calculate the derivative of m and b (formula obtained from calculus)
        md = (1/m) * np.sum(x * (y_predicted - y))
        bd = (1/m) * np.sum(y_predicted - y)

        # Adjust m and b using the formula
        m_curr = m_curr - learning_rate * md
        b_curr = b_curr - learning_rate * bd

        print(f"m: {m_curr}, b: {b_curr}, cost: {cost} iterations: {i}")
    return m_curr, b_curr

```

```python
import matplotlib.pyplot as plt
```

```python
m, b = gradient_descent(x,y)

plt.scatter(round(m),round(b))
plt.xlabel('X-Values')
plt.ylabel('Y-Values')
```
