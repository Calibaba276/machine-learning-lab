```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("./datasets/test_scores.csv")
data.head()
```

```python
math = data['math']
cs = data['cs']
```

```python
def fit(x,y):

    m_curr = b_curr = 0
    m = len(x)
    learning_rate = 0.0001
    iterations = 1000

    for i in range(iterations):
        y_predicted = m_curr * x + b_curr
        
        md = (1/m) * np.sum(x * (y_predicted - y))
        bd = (1/m) * np.sum(y_predicted - y)

        cost = (1/(2*m) * np.sum((y_predicted - y) ** 2))

        m_curr = m_curr - (learning_rate * md)
        b_curr = b_curr - (learning_rate * bd)

    return m_curr, b_curr
```

```python
def visualize(m, b, x, y):

    x_train = np.linspace(x.min(), x.max(), 100)
    y_train = ( m * x_train ) + b

    plt.scatter(x, y, label="Test Scores")
    plt.plot(x_train, y_train, label="Regression Line", color="gold")

    plt.xlabel('Math')
    plt.ylabel("Computer Science")
    plt.legend()
```

```python
m, b = fit(math, cs)
visualize(m,b, math, cs)
```
